```python
import torch

# 1. Set the device automatically
if torch.backends.mps.is_available():
    device = torch.device("mps")
elif torch.cuda.is_available():
    device = torch.device("cuda")
else:
    device = torch.device("cpu")

print(f"Using device: {device}")

# 2. Create a tensor and push it to your M1 GPU
x = torch.rand(3, 3)
x = x.to(device)

print("\nTensor successfully pushed to hardware:")
print(x)
```

# Tensor Operations:

```python
my_torch = torch.arange(10)
my_torch
```

```python
# Reshape and View
my_torch2 = my_torch.reshape(2,5)
my_torch2
```

```python
#reshape with unknow number of items
my_torch3 = my_torch.reshape(2, -1)
my_torch3
```

```python
# shared mutable state
my_torch[0] = 314

print(my_torch)
print(my_torch2)
print(my_torch3)
```

```python
#Slices
my_torch4 = torch.arange(10)
print(my_torch4[1]) #grab single item
print(my_torch4[0:5]) #grab slice
print()

my_torch5 = my_torch4.reshape(5, 2)
print(my_torch5)
print()
print(my_torch5[:,0])
```
